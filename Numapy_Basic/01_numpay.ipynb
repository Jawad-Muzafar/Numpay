# What is the Numpay?

>NumPy (Numerical Python) is an open source Python library that’s widely used in science and engineering. The NumPy library contains multidimensional array data structures, such as the homogeneous, N-dimensional ndarray, and a large library of functions that operate efficiently on these data structures.

# Why we use Numpay?

>- Python lists are excellent, general-purpose containers. They can be “heterogeneous”, meaning that they can contain elements of a variety of types, and they are quite fast when used to perform individual operations on a handful of elements.

>- Depending on the characteristics of the data and the types of operations that need to be performed, other containers may be more appropriate; by exploiting these characteristics, we can improve speed, reduce memory consumption, and offer a high-level syntax for performing a variety of common processing tasks. NumPy shines when there are large quantities of “homogeneous” (same-type) data to be processed on the CPU.

# What is array in Numpay?

>In computer programming, an array is a structure for storing and retrieving data. We often talk about an array as if it were a grid in space, with each cell storing one element of the data. For instance, if each element of the data were a number, we might visualize a “one-dimensional” array.

## Lets create an array

```python
#Array Creation
# You can create a NumPy array using the np.array() function. Here's an example:
a = np.array([1, 2, 3, 4, 5])
print(a)
```

## Types of the arrays

### 1D-array 

```python
#Creating a 1D array
b = np.array([10, 20, 30, 40, 50])
print(b)
```

### 2D-array

```python
#Creating a 2D array
C= np.array([[1, 2, 3], [4, 5, 6]])
print(C)
```

### 3D-array

```python
#Creating a 3D array
D = np.array([[[1, 2], [3, 4]], [[5, 6], [7, 8]]])
print(D)
```

```python
# We can verify the number of dimensions of an array using the .ndim attribute:
print(a.ndim)  # Output: 1
print(C.ndim)  # Output: 2
print(D.ndim)  # Output: 3
```

### Size of the array

```python
# Checking the size of an array using the .size attribute:
print(a.size)  # Output: 5
print(C.size)  # Output: 6
print(D.size)  # Output: 8
```

```python
#Length of an array using the len() function:
print(len(a))  # Output: 5
print(len(C))  # Output: 2 (number of rows)
print(len(D))  # Output: 2 (number of 2D arrays)
```

# Some basic arrays



```python
#Besides creating an array from a sequence of elements, you can easily create an array filled with 0’s:
zeros_array = np.zeros((3, 4))  # Creates a 3x4 array filled with zeros
print(zeros_array)
```

```python
#Or an array filled with 1’s:
ones_array = np.ones((2, 5))  # Creates a 2x5 array filled with ones
print(ones_array)
```

## We can create an array with a range of elements

```python
#An even an array that contains a range of evenly spaced intervals. To do this, you will specify the first number, last number, and the step size.
range_array = np.arange(0, 10, 2)  # Creates an array from 0 to 10 with a step of 2
print(range_array)
```

```python
#You can also use np.linspace() to create an array with values that are spaced linearly in a specified interval. This function takes the start and end values, as well as the number of samples to generate.
linspace_array = np.linspace(0, 1, 5)  # Creates an array of 5 values evenly spaced between 0 and 1
print(linspace_array)
```

## Adding and Sorting elements

```python
##Adding two arrays together
array1 = np.array([1, 2, 3])
array2 = np.array([4, 5, 6])
result = array1 + array2
print(result)  # Output: [5 7 9]
```

```python
## Sorting an array
unsorted_array = np.array([3, 1, 4, 1, 5, 9])
sorted_array = np.sort(unsorted_array)
print(sorted_array)  # Output: [1 1 3 4 5 9]
```

## Indexing in arrays

```python
#We can perfrom indexing and slicing on NumPy arrays similar to Python lists. Here's an example:
array = np.array([10, 20, 30, 40, 50])
print(array[0])      # Output: 10 (first element)
print(array[1:4])    # Output: [20 30 40] (elements from index 1 to 3)
print(array[-1])     # Output: 50 (last element)
```

## Converting 1D-array to 2D array

```python
#for converting 1D array to 2D array, we can use the reshape() method. Here's an example:
one_d_array = np.array([1, 2, 3, 4, 5, 6])
two_d_array = one_d_array.reshape(2, 3)  # Reshape to 2 rows and 3 columns
print(two_d_array)
```

```python
## We can convert a 1D array to a 2D array by adding a new axis using np.newaxis. Here's an example:
one_d_array = np.array([1, 2, 3, 4, 5, 6])
two_d_array = one_d_array[:, np.newaxis]  # Add a new axis to convert to 2D
print(two_d_array)
print(two_d_array.shape)  # Output: (6, 1) - 6 rows and 1 column
print (two_d_array.ndim)
```

## We can filter our data 

```python
## Filtration of an array
array = np.array([10, 15, 20, 25, 30])
filtered_array = array[array > 20]  # Filter elements greater than 20
print(filtered_array)  # Output: [25 30]
```
